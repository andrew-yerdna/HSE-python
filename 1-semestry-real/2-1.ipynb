```python
'''
Задание 1

Переведите содержимое файла purchase_log.txt в словарь purchases вида:
{'1840e0b9d4': 'Продукты', ...}

Пример работы программы при выводе первых двух элементов словаря purchases:

1840e0b9d4 'Продукты'
4e4f90fcfb 'Электроника'
'''
```

```python
import json

purchases = {}

with open('datasets/purchase_log.txt', 'r', encoding='utf-8') as file:
    next(file)
    for line in file:
        another_string = json.loads(line)

        key = another_string['user_id']
        value = another_string['category']
        purchases[key] = value

for key, value in list(purchases.items())[:2]:
    print(key, value)
```
